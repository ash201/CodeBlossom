```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, accuracy_score, confusion_matrix
```

```python
# 1. Linear Regression (predict Age)
# ===============================
# Drop rows with missing Age
df_lin = df[['Pclass', 'SibSp', 'Parch', 'Fare', 'Age']].dropna()

X_lin = df_lin[['Pclass', 'SibSp', 'Parch', 'Fare']]
y_lin = df_lin['Age']

X_train_lin, X_test_lin, y_train_lin, y_test_lin = train_test_split(X_lin, y_lin, test_size=0.2, random_state=42)

lin_reg = LinearRegression()
lin_reg.fit(X_train_lin, y_train_lin)
y_pred_lin = lin_reg.predict(X_test_lin)

mae = mean_absolute_error(y_test_lin, y_pred_lin)
print("Linear Regression - Mean Absolute Error:", mae)

```

```python
# 2. Logistic Regression (predict Survived)
# ===============================
# Select relevant features and drop missing
df_log = df[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Survived']].dropna()

# Convert categorical 'Sex' to numeric
df_log['Sex'] = df_log['Sex'].map({'male': 0, 'female': 1})

X_log = df_log[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']]
y_log = df_log['Survived']

X_train_log, X_test_log, y_train_log, y_test_log = train_test_split(X_log, y_log, test_size=0.2, random_state=42)

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_log, y_train_log)
y_pred_log = log_reg.predict(X_test_log)
```

```python
acc = accuracy_score(y_test_log, y_pred_log)
cm = confusion_matrix(y_test_log, y_pred_log)

print("Logistic Regression - Accuracy:", acc)
print("Logistic Regression - Confusion Matrix:\n", cm)
```
