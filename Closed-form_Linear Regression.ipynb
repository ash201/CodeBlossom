```python
import numpy as np 

X = np.array([[1], [2], [3], [4], [5]]) 
y = np.array([2, 4, 6, 8, 10]) 

X_b = np.c_[np.ones((X.shape[0], 1)), X] # shape becomes (5, 2) 

# Step 3: Apply the Normal Equation 
theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

 # Step 4: Extract weights 
b = theta[0] # intercept
w = theta[1] # slope 

print("Intercept (b):", b)
print("Slope (w):", w)
```

```python
import numpy as np

# Training data (y = 2x)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Add column of ones for bias term
X_b = np.c_[np.ones((X.shape[0], 1)), X]

# Normal Equation: θ = (XᵀX)^(-1) Xᵀy
theta = np.linalg.inv(X_b.T @ X_b) @ (X_b.T @ y)

print("Closed Form -> Intercept (b):", theta[0], "Slope (w):", theta[1])

# Prediction for x=6
X_new = np.array([[1, 6]])   # include the bias term
print("Prediction for x=6:", X_new @ theta)
```
