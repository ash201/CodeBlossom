```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = './Titanic-Dataset.csv'
df = pd.read_csv(path)
```

```python
df.head()
```

```python
df.head(10)
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['Fare'], alpha=0.5)
plt.title('Scatter plot of Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
pclass_counts = df['Pclass'].value_counts().sort_index()
plt.figure(figsize=(6, 4))
pclass_counts.plot(kind='bar')
plt.title('Bar chart of passenger count by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df['Age'].dropna().plot(kind='hist', bins=30, edgecolor='black')
plt.title('Histogram of Titanic passenger ages')
plt.xlabel('Age')
plt.show()
```

```python
plt.figure(figsize=(12, 4))
# Survival count plot
plt.subplot(1, 3, 1)
df['Survived'].value_counts().plot(kind='bar', color=['red', 'green'])
plt.title('Survival Count')
plt.xticks([0, 1], ['Not Survived', 'Survived'], rotation=0)
plt.ylabel('Count')

# Age distribution
plt.subplot(1, 3, 2)
df['Age'].plot(kind='hist', bins=30, color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')

# Survival rate by class
plt.subplot(1, 3, 3)
df.groupby('Pclass')['Survived'].mean().plot(kind='bar', color='orange', edgecolor='black')
plt.title('Survival Rate by Class')
plt.ylabel('Survival Rate')

plt.tight_layout()
plt.show()
```
